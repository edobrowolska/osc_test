```python
# BASIC INFORMATION ABOUT THE PROJECT: define id, title, description, status, and license for your project
# A custom id of the project, it can be related to the title, i.e. - 4datlantic-ohc. Use dash "-" symbol to separare words in the id"
project_id = "EO4CerealStress" 
# Specify the Title of your project. I.e. - 4DAtlantic-OHC. This should correspond to the title of the project as in the ESA contract. 
project_title = "EO4CerealStress:Response to Multiple Stressors" 
# A short description of the project:
project_description = "Cereal crops form the backbone of global food systems, yet their productivity is increasingly threatened by multiple stressors, including salinity, nutrient deficiencies, heavy metal toxicity, and drought. Understanding where, when, and how these stresses impact crops is essential for improving agricultural resilience and ensuring long-term food security.The EO4CerealStress project has developed a multi‑sensor, multi‑scale EO approach to monitor cereal crop responses to these stressors. By providing high spatial, spectral, and temporal resolution information, the project enables the detection of both individual and combined stress factors with high accuracy. Its workflow integrates Sentinel‑2, EnMAP, PRISMA, and hyperspectral drone data, combined with radiative transfer model (RTM) simulations, machine learning, deep learning, and extensive field validation campaigns conducted in 2023 and 2024. This synergistic use of data and modelling has significantly improved the detection and prediction of crop stress at field scale (10–30 m).The project has produced experimental EO‑based stressor products for drought, salinity, nutrient deficiency, heavy metal toxicity, and lodging across sites in Spain, Italy, and Austria. It has also developed new biophysical trait modelling frameworks leveraging Sentinel‑2, EnMAP, and hyperspectral drone data in Spain and Italy. AI‑ and ML‑driven pipelines have been created to predict drought stress from gap-filled Sentinel‑2 data in Austria and Canada, showcasing the scalability and transferability of the EO4CerealStress methodologies."
# Project status: pick from - ongoing or completed
project_status = "completed"

# Overall license for all related data that will be uploaded from the project., i.e. CC-BY-4.0. See the note in the markdown cell above to consult full list of available licenses. 
# If you have multiple licenses, you can pick 'various'
project_license = 'CC-BY-SA-4.0' 

# Define spatial extent of the project study area in epsg:4326
# if you have multiple disjoint study areas, specify the bounding box that covers all of them
# i.e project_s, project_w, project_n, project_e = -180.0, -90.0, 180.0, 90.0 
#project_s, project_w, project_n, project_e = -180.0, -90.0, 180.0, 90.0 

# 1. Define individual site bounding boxes [West, South, East, North]
site_spain = [-6.55, 36.80, -5.95, 37.25]
site_italy = [11.90, 44.80, 12.08, 44.95]
site_austria = [16.55, 48.20, 16.95, 48.45]
site_canada = [-99.40, 49.20, -96.70, 50.30]

# 2. Calculate the global bounding box (the Union of all sites)
# min_west, min_south, max_east, max_north
project_w = min(site_spain[0], site_italy[0], site_austria[0], site_canada[0])
project_s = min(site_spain[1], site_italy[1], site_austria[1], site_canada[1])
project_e = max(site_spain[2], site_italy[2], site_austria[2], site_canada[2])
project_n = max(site_spain[3], site_italy[3], site_austria[3], site_canada[3])
# The project start and end times
project_start_year, project_start_month, project_start_day = 2024, 8, 1
project_end_year, project_end_month, project_end_day = 2026,3,13

# Define the links to the project website and dedicated Project website on EO4SocietyLink. Discover the list of published projects here: https://eo4society.esa.int/projects/ 
website_link = "https://eo4cerealstress.org/"
eo4society_link = "https://eo4society.esa.int/projects/eo4cerealstress-theme-3-crop-response-to-multiple-stressors/"

# Define project themes, according to OSC ontology. Pick one or more from:
# - atmosphere, cryosphere, land, magnetosphere-ionosphere, oceans, solid-earth. 
# See the list here: https://github.com/ESA-EarthCODE/open-science-catalog-metadata/blob/main/themes/catalog.json
project_themes = ["land"]

# provide the Name and e-mail address to ESA Technical Officer (TO) supporting your project:
to_name = 'Zaib un nisa'    # Full Name and Surname
to_email = 'eo4cerealstress@soton.ac.uk'   

# List the consortium members in a tuple with format (name, contact_email), for example - ('University A', "contact@universitya.fr")
consortium_members = [('Prof. Roshanak Darvishzadeh', 'Prof. Andy Nelson', 'Dr Omid Ghorbanzadeh', 'Prof. Jadu Dash', 'Prof. Booker Ogutu', 'Prof. Aaron Berg', 'Dr Champagne Catherine', 'Dr Zaib un nisa', 'Dr Celik Furkan')]
```

```python
# combine the spatial and temporal extent
spatial_extent = pystac.SpatialExtent([[project_s, project_w, project_n, project_e]])
temporal_extent = pystac.TemporalExtent(
    [[datetime(project_start_year, project_start_month, project_start_day), 
      datetime(project_end_year, project_end_month, project_end_day)]])
project_extent = pystac.Extent(spatial=spatial_extent, temporal=temporal_extent)

# generate project collection
project_collection = create_project_collection(
    project_id, 
    project_title,
    project_description, 
    project_status,
    project_license,
    project_extent,
    project_themes,
    to_name,
    to_email,
    consortium_members,
    website_link,
    eo4society_link=eo4society_link
)

# validate the collection
project_collection.validate()

# Preview your project collection metadata: 
project_collection
# save the generated collection to a file(OPTIONAL)
project_collection.save_object(dest_href='./project_collection.json')
```
